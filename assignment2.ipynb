6.How can you iterate over a dictionary to access keys, values, and key-value pairs?

You can iterate over a dictionary using keys(), values(), and items().

keys() → returns all keys.
values() → returns all values.
items() → returns key-value pairs.

```python
student = {"name": "Rohit", "age": 28, "marks": 85.5}

# 1. Access keys
for key in student.keys():
    print(key)

# 2. Access values
for value in student.values():
    print(value)

# 3. Access key-value pairs
for key, value in student.items():
    print(key, value)
```

13.Group a list of words into lists of anagrams using a dictionary. Print the list of lists where each
inner list contains words that are anagrams of each other.
● Input: ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
● Anagram groups: [['listen', 'silent', 'enlist'], ['hello'], ['lemon', 'melon'], ['debitcard', 'badcredit']]

```python
words = ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]

anagrams = {}

for word in words:
    key = "".join(sorted(word))

    if key in anagrams:
        anagrams[key].append(word)
    else:
        anagrams[key] = [word]

print(list(anagrams.values()))
```

14.Merge two dictionaries. If there are common keys, sum their values. Print the merged
dictionary.

```python
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 5, "c": 15, "d": 40}

merged = dict1.copy()

for key, value in dict2.items():
    if key in merged:
        merged[key] += value
    else:
        merged[key] = value

print(merged)
```

16.Find mirror characters in a string using a dictionary.
For example, 'A' mirrors to 'Z', 'B' mirrors to 'Y',
Input string: "Hello World"

```python
mirror = {
    'A': 'Z', 'B': 'Y', 'C': 'X', 'D': 'W', 'E': 'V',
    'F': 'U', 'G': 'T', 'H': 'S', 'I': 'R', 'J': 'Q',
    'K': 'P', 'L': 'O', 'M': 'N', 'N': 'M', 'O': 'L',
    'P': 'K', 'Q': 'J', 'R': 'I', 'S': 'H', 'T': 'G',
    'U': 'F', 'V': 'E', 'W': 'D', 'X': 'C', 'Y': 'B',
    'Z': 'A'
}

word = "Hello World"
result = ""

for char in word:
    if char.upper() in mirror:
        result += mirror[char.upper()]
    else:
        result += char

print(result)
```

26.Find the minimum and maximum elements in a set without using the built-in min() and max()
functions.

```python
set1 = {10, 25, 7, 45, 32}

num_list = list(set1)

smallest = num_list[0]
largest = num_list[0]

for num in num_list:
    if num < smallest:
        smallest = num

    if num > largest:
        largest = num

print("Minimum:", smallest)
print("Maximum:", largest)
```

29.Find the symmetric difference of two sets with and without using built-in methods.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5, 6, 7}

result = set()

for num in set1:
    if num not in set2:
        result.add(num)

for num in set2:
    if num not in set1:
        result.add(num)

print(result)
```

30.Perform set operations (union, intersection, difference) on sets containing both integers and
strings without using built-in methods.
Set 1: {1, 2, 'apple', 'orange'}
Set 2: {2, 3, 'orange', 'banana'}

How it works:
- Union: Adds all elements from both sets, without duplicates.
- Intersection: Adds only elements present in both sets.
- Difference: Adds elements present in set1 but absent from set2.

```python
set1 = {1, 2, 'apple', 'orange'}
set2 = {2, 3, 'orange', 'banana'}

# Union
union_set = set1.copy()

for item in set2:
    union_set.add(item)

print("Union:", union_set)

# Intersection
intersection_set = set()

for item in set1:
    if item in set2:
        intersection_set.add(item)

print("Intersection:", intersection_set)

# Difference (Set 1 - Set 2)
difference_set = set()

for item in set1:
    if item not in set2:
        difference_set.add(item)

print("Difference:", difference_set)
```
