27.How to sort a list in ascending and descending order without using the sort function?

```python
list1 = [5, 2, 8, 1, 3]

ascending = list1.copy()
descending = list1.copy()

for i in range(len(ascending)):
    for j in range(i + 1, len(ascending)):
        if ascending[i] > ascending[j]:
            ascending[i], ascending[j] = ascending[j], ascending[i]

for i in range(len(descending)):
    for j in range(i + 1, len(descending)):
        if descending[i] < descending[j]:
            descending[i], descending[j] = descending[j], descending[i]

print("Ascending:", ascending)
print("Descending:", descending)
```

28.How to sort a tuple?

Since tuples are immutable in Python, you cannot sort a tuple directly. You can use sorted() to sort it, which returns a list.

```python
tuple1 = (5, 2, 8, 1, 3)

ascending = tuple(sorted(tuple1))
descending = tuple(sorted(tuple1, reverse=True))

print("Ascending:", ascending)
print("Descending:", descending)
```

remove() deletes an element by its value, whereas pop() deletes an element by its index and returns the removed element.


```python
# Using remove()
list1 = [10, 20, 30, 40]
list1.remove(20)
print(list1)  # [10, 30, 40]

# Using pop()
list2 = [10, 20, 30, 40]
x = list2.pop(1)
print(list2)  # [10, 30, 40]
print(x)      # 20
```

33.What is the difference between sort and sorted?

sort() modifies the original list, whereas sorted() returns a new sorted list without changing the original.

```python
list1 = [3, 1, 2]

list1.sort()
print(list1)  # [1, 2, 3]

list2 = [3, 1, 2]

result = sorted(list2)
print(result) # [1, 2, 3]
print(list2)  # [3, 1, 2]
```

34.What is the difference between reverse and reversed function?

reverse() modifies the original list, whereas reversed() returns an iterator that lets you access the elements in reverse order without changing the original list.

```python
list1 = [1, 2, 3, 4]

list1.reverse()
print(list1)  # [4, 3, 2, 1]

list2 = [1, 2, 3, 4]

print(list(reversed(list2)))  # [4, 3, 2, 1]
print(list2)                 # [1, 2, 3, 4]
```

35.What is the difference between copy and deep copy?

copy() creates a shallow copy, so nested objects are still shared; deepcopy() creates an independent copy of nested objects as well.

```python
import copy

list1 = [[1, 2], [3, 4]]

# Shallow copy
list2 = list1.copy()
list2[0][0] = 100
print(list1)  # [[100, 2], [3, 4]]

# Deep copy
list3 = copy.deepcopy(list1)
list3[0][0] = 500
print(list1)  # [[100, 2], [3, 4]]
print(list3)  # [[500, 2], [3, 4]]
```

38.Given a list of integers, create a new list where each element is replaced by the next largest
number in the original list. If there is no larger number, replace the element with -1.
● input_list = [4, 2, 6, 8, 1, 2]
● output_list = [6, 6, 8, -1, 2, -1]

```python
input_list = [4, 2, 6, 8, 1, 2]
output_list = []

for num in input_list:
    next_largest = -1

    for n in input_list:
        if n > num:
            if next_largest == -1 or n < next_largest:
                next_largest = n

    output_list.append(next_largest)

print(output_list)
```

39.Write a function to reverse the elements within each tuple in a list of tuples.
● tuples = [(1, 2), (3, 4), (5, 6)]
● Output: [(2, 1), (4, 3), (6, 5)]

```python
def reverse_tuples(tuples):
    result = []

    for item in tuples:
        result.append(item[::-1])

    return result


tuples = [(1, 2), (3, 4), (5, 6)]

print(reverse_tuples(tuples))
```

40.Write a function to add corresponding elements of tuples in two lists of tuples
● list1 = [(1, 2), (3, 4)]
● list2 = [(5, 6), (7, 8)]
● Output: [(6, 8), (10, 12)]

```python
def add_tuples(list1, list2):
    result = []

    for i in range(len(list1)):
        result.append((list1[i][0] + list2[i][0],
                       list1[i][1] + list2[i][1]))

    return result


list1 = [(1, 2), (3, 4)]
list2 = [(5, 6), (7, 8)]

print(add_tuples(list1, list2))
```
